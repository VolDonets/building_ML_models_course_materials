# Task description

## Overall description

In this lab work, you will transition from tabular Reinforcement Learning to Deep Reinforcement Learning by exploring Deep Q-Networks (DQN). You will learn how neural networks can be used as function approximators to estimate Q-values, allowing agents to solve environments with infinite or highly complex continuous state spaces.

You will start by analyzing a baseline DQN implementation for a classic control problem (CartPole) and then scale up to train an agent in a complex visual environment (an Atari game).
